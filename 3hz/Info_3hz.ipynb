```python
import numpy as np
import matplotlib.pyplot as plt
test_path = "../../Datasets/3hz/Multi/test_set.npz"
train_path = "../../Datasets/3hz/Multi/train_set.npz"
val_path = "../../Datasets/3hz/Multi/valid_set.npz"
def inspect_npz(file_path):
    try:
        data = np.load(file_path)
        for key in data:
            print(f"{key}: shape {data[key].shape}")
            # plt.imshow(data[key][1], cmap= 'gray')
            # plt.show()
        data.close()
    except Exception as e:
        print(f"Error loading {file_path}: {e}")

print("Contents of test_set.npz:")
inspect_npz(test_path)
print("\nContents of train_set.npz:")
inspect_npz(train_path)
print("\nContents of valid_set.npz:")
inspect_npz(val_path)
```

```python
data = np.load(test_path)
import cv2
```

```python
image = data['arr_0'][0]
image = (image - np.min(image)) / (np.max(image) - np.min(image))
resized_image = cv2.resize(image, (256, 256), interpolation=cv2.INTER_LINEAR)
image1 = data['arr_1'][0]
image1 = (image1 - np.min(image1)) / (np.max(image1) - np.min(image1))
resized_image1 = cv2.resize(image1, (256, 256), interpolation=cv2.INTER_LINEAR)

plt.subplot(2, 2, 1)
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.title('X Original')

plt.subplot(2, 2, 2)
plt.imshow(resized_image, cmap='gray')
plt.axis('off')
plt.title('X Resized')

plt.subplot(2, 2, 3)
plt.imshow(image1, cmap='gray')
plt.axis('off')
plt.title('y Original')

plt.subplot(2, 2, 4)
plt.imshow(resized_image1, cmap='gray')
plt.axis('off')
plt.title('y Resized')

plt.tight_layout()
plt.show()
```
