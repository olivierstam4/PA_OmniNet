```python
import numpy as np
import cv2
import os

def inspect_npz(file_path):
    try:
        data = np.load(file_path)
        print(f"Inspecting {file_path}:")
        for key in data:
            print(f"{key}: shape {data[key].shape}, dtype {data[key].dtype}")
    except Exception as e:
        print(f"Error loading {file_path}: {e}")

def resize(file_paths):
    for file_path in file_paths:
        data = np.load(file_path)
        resized_data = {}
        for key in data:
            original = data[key]
            resized = np.zeros((original.shape[0], 256, 256), dtype=original.dtype)
            for i in range(original.shape[0]):
                resized[i] = cv2.resize(original[i], (256, 256), interpolation=cv2.INTER_LINEAR)
            resized_data[key] = resized

        new_file_path = os.path.splitext(file_path)[0] + "_256.npz"
        np.savez(new_file_path, **resized_data)

        print(f"Processed {file_path}")

hz = [
    # "../Datasets/3hz/Multi/test_set.npz",
    # "../Datasets/3hz/Multi/train_set.npz",
    # "../Datasets/3hz/Multi/valid_set.npz",
    # "../Datasets/3hz/Single/test_set.npz",
    # "../Datasets/3hz/Single/train_set.npz",
    "../Datasets/3hz/Single/valid_set.npz",
]
for file_path in hz:
    inspect_npz(file_path)
resize(hz)
```
